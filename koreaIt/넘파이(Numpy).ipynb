```python
import numpy as np

n1 = np.sum([1, 2, 3])
n1
```

```python
x = np.array([1, 2, 3])
print(x)
print(x.shape)

print(x[0])
print(x[2])
print(x[-1])
print(x[0:3])
print(x[:])
print(x[:-1])
```

```python
d2 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(d2)
print(d2.shape)

print(d2[0, 0])
print(d2[1, 1])
print(d2[0:, 0:])
print(d2[1:, 1:])
```

```python
a = np.arange(10)
print(a)
b = np.arange(0, 10, 1)
print(b)
c = np.arange(0, 10, 2)
print(c)

a2 = a.reshape(2,5)
print(a2)

a3 = a2.flatten()
print(a3)
```
