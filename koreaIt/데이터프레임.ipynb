```python
# 데이터 프레임 직접 생성
import pandas as pd

data = [
    ['키보드', 20000],
    ['마우스', 32000],
    ['USB 메모리', 11000]
]

df = pd.DataFrame(data, columns=['name', 'price'])
df

df['name']
df[['name', 'price']]
```

```python
import pandas as pd

df = pd.DataFrame({
    "name": ['키보드', '마우스', 'USB 메모리'],
    "price": [20000, 32000, 11000]
})

# 칼럼 추가
df['spec'] = ['유선', '무선', '128GB']

df['할인가'] = df['price'] * 0.8
df

df = df.rename(columns={"name": "품명", "price":"가격", "spec":"제품상세"})
df
```

```python
# loc[행]
df.loc[0]
df.loc[1]
df.loc[0:1]
df.loc[[0, 1]]

# loc[행, 열]
df.loc[0, "품명"]
df.loc[1, "가격"]
df.loc[0:1, "품명"]
df.loc[0:1, ["품명", "가격"]]
```

```python
# iloc[행인덱스, 열인덱스]
df.iloc[0, 0]
df.iloc[1, 1]
df.iloc[0:2, 0]
df.iloc[0:2, 0:2]
# df.iloc[:, :]
```

```python
df2 = df.copy()
df2

df2 = df2.set_index("품명")
df2
df2.loc["키보드"]
df2.loc[["키보드", "마우스"]]
df2.loc["키보드":"USB 메모리"]
```
