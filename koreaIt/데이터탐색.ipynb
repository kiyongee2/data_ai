```python
# 데이터 프레임 만들기
import pandas as pd

food = pd.DataFrame({
    "메뉴": ["김밥", "비빔밥", "자장면", "우동", "김밥", "자장면", "김밥"],
    "가격": [3000, 7000, 5500, 5500, 3500, 6000, 4000],
    "분류": ["한식", "한식", "중식", "일식", "한식", "중식", "한식"]
})
food

food.to_csv("datas/food.csv", index=False)

df = pd.read_csv("datas/food.csv")
df
```

```python
df.head()
df.tail()
df.info()

df.value_counts()
df["메뉴"].value_counts()
df["분류"].value_counts()

df.describe()
df.describe(include='object')
# df.describe(include='number')
```

```python
# 칼럼 추가
discount = 0.1
df["할인가"] = df["가격"] * (1 - discount)
df
df.info()
   
# 자료형 변환 - astype()
df["할인가"] = df["할인가"].astype("int64")
df
# df.info()
```

```python
# 조건 검색
result1 = (df["메뉴"] == "자장면")
result1
df[result1]

result4 = (df["메뉴"] == "자장면") & (df["가격"] >= 6000)
df[result4]
```

```python
# 정렬
df.sort_values("가격", ascending=False)
```

```python
# 판다스의 통계, 수학 함수
df["가격"].count()
df["가격"].sum()
round(df["가격"].mean(), 2)
df["가격"].var()
df["가격"].std()
df["가격"].min()
df["가격"].idxmax()

df["가격"].quantile(.25)
df["가격"].quantile(.50)

result = df["가격"] < df["가격"].quantile(.25)
print(df[result])

df["가격"].mode()[0]
```

```python
import pandas as pd

# "수량" 칼럼 추가
df["수량"] = [10, 10, 15, 5, 7, 7, 8]
df

# csv 파일로 변환
df.to_csv("./datas/food2.csv", index=False)

# csv 파일 읽기
food = pd.read_csv("./datas/food2.csv")
food
```

```python
# 그룹핑 - groupby()
df["분류"].value_counts()

food.groupby("메뉴").mean(numeric_only=True)
food.groupby("분류").mean(numeric_only=True)

food.groupby("분류")["가격"].mean()
food.groupby("분류")["수량"].mean()
```
