```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# 2번
abandon_df = pd.read_csv("./datasets/cart_abandon.csv")
abandon_df
```

```python
# 3번
abandon_df.groupby('user_type')['time_on_site'].mean()
```

```python
# 회원 유형에 따른 웹사이트 체류 시간 시각화 
sns.boxplot(abandon_df, x="user_type", y="time_on_site")
답04 = 2
```

```python
# 4번
abandon_df.isnull().sum()

median = abandon_df['time_on_site'].median()
print(median) # 8.094999999999999
mean = abandon_df['cart_value'].mean()
print(mean) #262.1885336538462
# abandon_df['time_on_site'].fillna(median, inplace=True)
# inplace=True 경고 발생
abandon_df['time_on_site'] = abandon_df['time_on_site'].fillna(median)
abandon_df['cart_value'] = abandon_df['cart_value'].fillna(mean)
abandon_df.isnull().sum()
```

```python
abandon_df.info()
```

```python
# 5번
# 형변환(object -> float)
abandon_df["prev_purchase"] = abandon_df["prev_purchase"].map({"Y":1, "N":0})
abandon_df["discount_applied"] = abandon_df["discount_applied"].map({"Y":1, "N":0})
abandon_df["user_type"] = abandon_df["user_type"].map({"Y":1, "N":0})
abandon_df.info()
abandon_df.head()
```

```python
# 원-핫 인코딩
cat_col = ["device", "country", "payment_method"]
incoding_df = pd.get_dummies(abandon_df, columns=cat_col, drop_first=True)
incoding_df.info()
incoding_df.head()
```

```python
# 6번 num_items 칼럼 삭제 
# incoding_df.drop(columns='num_items', axis=1, inplace=True)
incoding_df.drop('num_items', axis=1, inplace=True)
incoding_df.info()
```

```python
# 7번
result = incoding_df['time_on_site'] < 0
print(result.sum()) #4개
print(incoding_df[result].index) #[137, 192, 774, 1294]
filtering_df = incoding_df.drop(incoding_df[result].index, axis=0)
# filtering_df.tail(10)
# 삭제후 index 재설정됨
filtering_df.reset_index(drop=True, inplace=True)
filtering_df.info()
filtering_df.tail(10)
```

```python
# 8번. 데이터 분리 - 변수 선택 
X = filtering_df.drop(columns='cart_abandon_yn', axis=1)
y = filtering_df['cart_abandon_yn']
# X.info()
y
```

```python
# 9번
# pip install scikit-learn
from sklearn.model_selection import train_test_split

X_train, X_valid, y_train, y_valid = train_test_split(X, y,  
        test_size=0.2, random_state=11, stratify=y)
```

```python
# 10번 데이터 표준화 - 스케일링 
from sklearn.preprocessing import StandardScaler

ss = StandardScaler()
X_train = ss.fit_transform(X_train)
X_test = ss.transform(X_valid)
```

```python
# 11. 랜덤포레스트 분류 모델 
from sklearn.ensemble import RandomForestClassifier

rfc = RandomForestClassifier(n_estimators=100, max_depth=7, random_state=42)
rfc.fit(X_train, y_train) # 모델 학습
```

```python
# 12 XGBClassifier 모델 
from xgboost import XGBClassifier

xgbc = XGBClassifier(n_estimators=100, max_depth=5, learning_rate=0.1, random_state=42)
xgbc.fit(X_train, y_train)
```

```python
# 13
from sklearn.metrics import accuracy_score, f1_score

rfc_predict = rfc.predict(X_test)
xgbc_predict = xgbc.predict(X_test)

print("rfc_accuracy_score:", accuracy_score(y_valid, rfc_predict))
print("rfc_fl_score:", f1_score(y_valid, rfc_predict))

print("xgbc_accuracy_score:", accuracy_score(y_valid, xgbc_predict))
print("xgbc_fl_score:", f1_score(y_valid, xgbc_predict))
```

```python
# 14
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.utils import to_categorical

tf.random.set_seed(42)
y_train = to_categorical(y_train, num_classes=2)
y_valid = to_categorical(y_valid, num_classes=2)
```

```python
model = Sequential()
model.add(Dense(64, activation= 'relu', input_shape=(X_train.shape[1],)))
model.add(Dense(32, activation='relu'))
model.add(Dense(2, activation='softmax'))

model.compile(optimizer = 'adam',
              loss = 'categorical_crossentropy',
              metrics = ['accuracy'])

model.fit(X_train, y_train, epochs=50, batch_size = 128,
          validation_data = (X_test, y_valid))
```
