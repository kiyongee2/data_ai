```python
# 데이터프레임 생성 2
import pandas as pd
# data - 딕셔너리

df = pd.DataFrame({
    "name": ["키보드", "마우스", "USB 메모리"],
    "price": [20000, 32000, 11000]
})
print(df)

# 칼럼 추가
df["spec"] = ["유선", "무선", "128GB"]
print(df)

# 할인가격 - 20% 할인 
df["할인가"] = df["price"] * (1 - 0.2)
print(df)

# 칼럼명 변경 - rename()
df = df.rename(columns = {"name": "품명", "price": "가격", "spec": "제품상세"})
print(df)
```

```python
# 행 선택 - loc, iloc 속성 
# loc[행(행값)]
print(df.loc[0])
print(df.loc[1])
print(df.loc[[0, 1]]) #인덱싱(리스트)
print(df.loc[0:1])   #슬라이싱(범위연산자)

# loc[행(행값), 열(칼럼명)]
print(df.loc[0, "품명"])
print(df.loc[1, "가격"])
print(df.loc[0:1, "품명"])
print(df.loc[0:1, ["품명", "가격"]])
print(df.loc[0:1, "품명" : "제품상세"])
print(df.loc[:, "품명":"할인가"])  #행, 열 전체

# iloc[행(인덱스), 열(인덱스)], 인덱스 이므로 종료값-1로 함
print(df.iloc[0, 0])
print(df.iloc[1, 1])
print(df.iloc[0:2, 0])
print(df.iloc[0:2, 0:2])
print(df.iloc[:, :]) #행, 열 전체
```

```python
# 데이터 프레임 복사 - copy()
df2 = df.copy()
print(df2)

# 인덱스 설정 - set_index(칼럼명)
df2 = df2.set_index("품명")
print(df2)

print("===========================================")
# 행 선택(loc)
print(df2.loc["키보드"])
print(df2.loc[["키보드","마우스"]])
print(df2.loc["키보드":"USB 메모리"])

# loc[행(행값), 열(칼럼명)] 
print(df2.loc["키보드", "가격"])
print(df2.loc["키보드", "가격":"제품상세"])
print(df2.loc["키보드", ["가격", "제품상세"]])

# iloc[행(인덱스), 열(인덱스)]
print(df.iloc[0, 0])
print(df.iloc[1, 1])
```

```python
df3 = df.copy()
# print(df3)

# 요소값 변경
# df3.loc[2, "가격"] = 12000
# df3.loc[2, "할인가"] = 9600.0
df3.iloc[2, 1] = 12000
df3.iloc[2, 3] = df3.iloc[2, 1] * 0.8
# print(df3)

# 행 추가 
df3.loc[3] = ["키보드", 35000, "무선", 0.0]
df3.loc[3, "할인가"] = df3.loc[3, "가격"] * 0.8
print(df3)

# 열 추가(칼럼 추가)
df3["수량"] = 100
print(df3)

# 행 삭제 - drop() axis=0은 행삭제, axis=1은 열 삭제
df3 = df3.drop(3, axis=0) 
print(df3)

# 열 삭제
# df3 = df3.drop("할인가", axis=1)
df3 = df3.drop(["제품상세","할인가"], axis=1)
print(df3)
```

```python
df4 = df.copy()
print(df4)

# 칼럼 검색
print(df4["품명"])
print(df4[["품명", "가격", "제품상세"]])

# 조건 검색(필터링) 
# 품명이 '마우스'인 제품의 정보
result1 = (df4["품명"] == "마우스")
print(df4[result1])

# 가격이 30000원을 초과하는 제품의 정보 
result2 = (df4["가격"] > 30000)
print(df4[result2])

# 조건이 2개 이상이면 논리 연산 (&, |) - 조건은 반드시 괄호로 묶음 
# result3 = (df4["품명"] == "마우스") & (df4["가격"] >= 20000)
result3 = (df4["품명"] == "마우스") | (df4["가격"] >= 20000)
print(df4[result3])

# 품명이 'USB 메모리'가 아닌 제품
result4 = ~(df4["품명"] == "USB 메모리")
print(df4[result4])
```

```python
import pandas as pd

df = pd.read_csv("food.csv")
df

# 조건 검색(필터링)
# 메뉴가 '자장면'인 음식의 정보 
result1 = (df["메뉴"] == '자장면')
print(df[result1])

# 메뉴가 '김밥'이 아닌 음식의 정보
result2 = (df["메뉴"] != '김밥')
# result2 = ~(df["메뉴"] == '김밥')
print(df[result2])

# 가격이 4000원 미만인 음식의 정보
result3 = (df["가격"] < 4000)
print(df[result3])

# 메뉴가 '자장면'이고, 가격이 6000원 이상인 음식의 정보
# result4 = (df["메뉴"] == '자장면') & (df["가격"] >= 6000)
result4 = (df["메뉴"] == '자장면') | (df["가격"] >= 6000)
print(df[result4])
```

```python
# 수학, 통계 함수 - len(), sum(), mean(), std(), 
import pandas as pd

df = pd.read_csv("food.csv")
df

# 개수
print(len(df))

# 반올림 함수
print(round(23.61, 1))

# 판다스의 통계 함수
print("개수:", df["가격"].count())
print("합계:", df["가격"].sum())
print("평균:", round(df["가격"].mean(), 2))
print("분산:", round(df["가격"].var(), 2))
print("표준편차:", round(df["가격"].std(), 2))
print("최대값:", df["가격"].max())
print("최소값:", df["가격"].min())
print("최대값의 위치:", df["가격"].idxmax())
print("최소값의 위치:", df["가격"].idxmin())
df

# 사분위수 - quantile()
print("1사분위수:", df["가격"].quantile(.25))
print("2사분위수:", df["가격"].quantile(.50))
print("3사분위수:", df["가격"].quantile(.75))
print("4사분위수:", df["가격"].quantile(1.0))

# 조건 - 1사분위수 보다 작은 가격을 출력 
result = df["가격"] < df["가격"].quantile(.25)
print(df[result])

# 최빈값 - mode()
print("최빈값:", df["가격"].mode()[0])

# nlargest(), nsmallest()
print(df.nlargest(2, "가격"))
print(df.nsmallest(2, "가격"))
```

```python
# 데이터프레임 만들기
import pandas as pd

data = {
    "메뉴": ["김밥", "비빔밥", "자장면", "우동", None, "자장면", "김밥"],
    "가격": [3000, 7000, 5500, None, 3500, 6000, 4000],
    "분류": ["한식", "한식", "중식", "일식", "한식", "중식", "한식"]
}

food = pd.DataFrame(data)
food
food.info()

# 결측치 확인 - isnull()
print(food.isnull())
print(food.isnull().sum()) #True(1) / False(0)

# 결측치 대체
# 수치형 - 중간값으로 대체 
median = food["가격"].median()
median

# fillna()
food["가격"] = food["가격"].fillna(median)
food

# 문자형 - 최빈값 
frequency = food["메뉴"].mode()[0]
frequency
food["메뉴"] = food["메뉴"].fillna(frequency)
food

print(food.isnull().sum())
```

```python
# csv 파일 읽기
food = pd.read_csv("food2.csv")
food

# 그룹화
# food.groupby("메뉴").mean(numeric_only=True)
food.groupby("분류").mean(numeric_only=True) #수치형 칼럼 평균 

food.groupby("분류")["가격"].mean() # 가격의 평균 
food.groupby("분류")["수량"].mean() # 수량의 평균 

# agg() - 여러개의 칼럼에 대해 집계 연산을 동시에 수행
food.groupby(["분류", "메뉴"]).agg(['mean'])
food.groupby(["분류", "메뉴"]).agg(['mean', 'sum'])
```
