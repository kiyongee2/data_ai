```python
import pandas as pd

df = pd.DataFrame({
    "name": ['키보드', '마우스', 'usb '],
    "price": [17000, 25000, 78000],
    "spec": ['유선', None, '22inch']
})
# print(df)

# 칼럼 검색
df['name']
df[['name', 'price']]

# head(), tail() - 기본 5개
df.head(2)
df.tail(2)

# 데이터 정보 - info()
df.info()

# 결측치 - isnull(), 개수 - sum()
df.isnull()
df.isnull().sum()  # True - 1개

# 통계 요약 - describe(), 생략 가능-include='number'
df.describe(include='number')
df.describe(include='object')

# 칼럼 이름 변경 - rename()
df = df.rename(columns={"name": "이름", "price": "가격", "spec": "스펙"})
df
```
