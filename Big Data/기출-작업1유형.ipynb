### 6회 기출문제

##### 문제1) 출동시간과 도착시간 차이가 평균적으로 가장 오래 걸린 소방서의 <br><br>시간을 분으로 변환해 출력하시오

```python
import pandas as pd

df = pd.read_csv("data6-1.csv")
# df
# df.info()

# datetime 변경
df['출동시간'] = pd.to_datetime(df['출동시간'])
df['도착시간'] = pd.to_datetime(df['도착시간'])
# df.info()

# 출동시간과 도착시간의 차이: 총 초로 변환 - dt.total_seconds()
df["시간차이(분)"] = (df["도착시간"] - df["출동시간"]).dt.total_seconds() / 60
print(df)

# 소방서별 평균 시간 차이
mean_diff = df.groupby("소방서")["시간차이(분)"].mean()
print(mean_diff)

# 평균 차이가 가장 큰 소방서 
idx = mean_diff.idxmax()  #최대값의 위치 
print(round(mean_diff[idx]))
```

##### 문제2) 학교에서 교사 한 명당 맡은 학생 수가 가장 많은 학교를 찾고, <br><br>그 학교의 전체 교사수를 구하시오

```python
import pandas as pd

df = pd.read_csv("data6-2.csv")
df

# 총학생수 
df["총학생수"] = df["1학년"] + df["2학년"] + df["3학년"] + df["4학년"] + df["5학년"] + df["6학년"]
# df["총학생수"] = df.iloc[:, 2:].sum(axis=1) #axis=1 열의 합계 

# "교사1인당 학생수"
df["교사1인당 학생수"] = df["총학생수"] / df["교사수"]
df

# "교사1인당 학생수" 내림차순 
df = df.sort_values("교사1인당 학생수", ascending=False)
print(df)

# 맨위 행의 교사수
print(df.iloc[0, 1])
```

##### 문제3) 연도별로 총 범죄 건수(범죄 유형별 발생 건수의 합)의 월평균 값을 
##### 구한후, 그 값이 가잔 큰 연도를 찾아 해당 연도의 총 범죄 건수의 
##### 월평균 값을 출력하시오. (반올림하여 정수 출력)

```python
import pandas as pd

df = pd.read_csv("data6-3.csv")
print(df)
df.info()

# 총 범죄 건수
df["총범죄건수"] = df.iloc[:, 1:7].sum(axis=1)

# '연도' 칼럼 
df["연도"] = df["날짜"].str[:4]
df

# 연도별 총 범죄건수의 월 평균 
month_mean = df.groupby("연도")["총범죄건수"].sum() / 12

# 가장 큰 연도
idx = month_mean.idxmax()
print(round(month_mean[idx]))
```
