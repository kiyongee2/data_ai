### matplotlib 라이브러리

 - Python에서 가장 널리 사용되는 시각화 라이브러리이다. <br>
 - 데이터 분석, 과학 시각화, 보고서 작성 등 다양한 분야에서 그래프를 그릴 때 사용됨

```python
# 데이터 시각화
import matplotlib.pyplot as plt

plt.rc('font', family='Malgun Gothic')  #한글 글꼴일때 명시
plt.title('기본 그래프 그리기 ')
# plt.plot([10, 20, 30, 40])  #리스트 1개 y축값(결과값)
plt.plot([1, 2, 3, 4], [10, 40, 20, 30]) #리스트 2개, x축값, y축값
plt.show()
```

```python
import matplotlib.pyplot as plt

# 데이터 준비
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

# 선 그래프 그리기
plt.plot(x, y, label='Prime numbers', color='blue', marker='o')
plt.title('Line Graph')
plt.legend() #범례
plt.show()
```

```python
import matplotlib.pyplot as plt

fruits = ['Apple', 'Banana', 'Grape', 'Orange']
counts = [10, 15, 7, 12]

plt.rc('font', family="Malgun Gothic")  #한글 글꼴일때 명시
plt.bar(fruits, counts, color='yellow')
plt.title('과일 수량')
plt.xlabel('과일')
plt.ylabel('수량')
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.rc('font', family='Malgun Gothic')  #한글 글꼴일때 명시
plt.title('범례 및 색상 표시하기') #제목
plt.plot([10, 20, 30, 40], label='asc', color='r') # 범례(label), red = r
plt.plot([40, 30, 20, 10], label='desc', color='g')
plt.legend() # 범례 표시
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.rc('font', family='Malgun Gothic')  #한글 글꼴일때 명시
plt.title('x축 제목, y축 제목 표시') #제목
plt.plot([40, 30, 20, 10])
plt.xlabel("인구")
plt.ylabel("식량")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.plot([10, 20, 30, 40], 'b.', label='circle') # 범례(label), red = r
plt.plot([40, 30, 20, 10], 'r^', label='triangle')
plt.legend() # 범례 표시
plt.show()
```

```python
# 막대 그래프
import matplotlib.pyplot as plt

print(range(6))
print(list(range(6)))

# bar(막대를 표시할 위치, 막대의 높이)
# plt.bar([1, 2, 3, 4], [6, 3, 2, 4])
plt.bar(range(6), [1, 2, 3, 5, 6, 7])
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

data = np.random.randn(1000)  # 정규분포 데이터 생성

plt.hist(data, bins=30, color='skyblue', edgecolor='black')
plt.title('Histogram Example')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

```python
import matplotlib.pyplot as plt
import random

numbers = []
for i in range(10):
    dice = random.randint(1, 6)
    numbers.append(dice)

print(numbers)

plt.rc('font', family='Gulim') 
plt.title('주사위 던지기') 
plt.hist(numbers, bins=6) # bins는 가로축의 구간 개수
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.figure(figsize=(10, 4))

# 1행 2열 중 첫 번째
plt.subplot(1, 2, 1)
plt.plot(x, y1)
plt.title('Sine')

# 1행 2열 중 두 번째
plt.subplot(1, 2, 2)
plt.plot(x, y2)
plt.title('Cosine')

plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 예제 데이터셋 로드
df = sns.load_dataset('titanic')

# 성별에 따른 생존자 수 시각화 (막대그래프)
plt.rc('font', family='Malgun Gothic')  #한글 글꼴일때 명시
sns.countplot(data=df, x='sex', hue='survived')

plt.title('성별에 따른 생존자 수')
plt.show()
```
